# 페이징 처리
-  대량의 데이터를 여러 페이지로 나눠서 출력하는 것.
-  Django에서는 Paginator와 Page 클래스를 통해 처리한다.
  
## Paginator 클래스
- 전체 페이징 처리를 관리하는 클래스
- 전체 데이터관련 정보, 각 페이지당 보여줄 데이터의 정보 등을 제공

## Page 클래스
- 한페이지에대한 데이터를 관리
- Paginator를 통해서 생성.
    - `Pagenator객체.page(페이지 번호)`
- iterable 타입. 페이지에 속한 데이터들을 제공
- Page객체.object_list 속성: 페이지가 가진 데이터들을 List로 반환

```python
txt = "0123456789abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ"
dataset = list(txt)
len(txt), len(dataset)
```

```python
from django.core.paginator import Paginator

# page 관리 -> Paginator 객체를 생성 -: 전체 data다 넣어서 생성
pn = Paginator(dataset, 5)	# (전체 data, page당 데이터 개수)
```

```python
print("전체 data 수 :", pn.count)
print("총 page 수 :", pn.num_pages)
print("시작 page 번호, 끝 page 번호 :", pn.page_range)
```

```python
for page in pn.page_range:
    print(page, end = "\t")
```

```python
# 특정 page 값 조회 : Page객체 반환
page1 = pn.page(1)
page1
```

```python
page13 = pn.page(13)
```

```python
pn.page(100)	# 없는 page, EmptyPage Exception 발생
```

```python
# Page객체가 가진 data를 조회 -> Page: Iterable, Subscriptable
# page1[1]
for data in page1:
    print(data, end = ", ")
for data in page13:
    print(data, end = ", ")
```

```python
# page의 data를 list로 반환
page1.object_list
```

## 이전/다음 페이지가 있는지 
- `Page객체.has_previous()` / `Page객체.has_next()`
- 1페이지: 이전페이지? X, 다음페이지? O
- 중간 페이지: 이전페이지? O, 다음페이지? O
- 마지막 페이지: 이전? O, 다음페이지? X

```python
page1.has_previous(), page1.has_next()
```

```python
page13.has_previous(), page13.has_next()
```

## 이전/다음페이지 번호 조회
- **Page객체.number:** 현재 페이지 번호
- **Page객체.previous_page_number():** 이전페이지 번호 조회
- **page객체.next_page_number():** 다음페이지 번호 조회

```python
page13.number, page13.previous_page_number(), page1.next_page_number()
```

```python
page13.next_page_number()	# 다음 page가 없을 경우 EmptyPage Exception
```

## 각 페이지별 데이터를 출력(조회)

```python
# 한 page당 K개의 data 출력
# ------K page------
# a, b, c, d, e, ...
paginate_by = 10	# k개
```

```python
# Paginator객체 생성
pn = Paginator(dataset, paginate_by)
pn.page_range
```

```python
# loop로 각 page의 data 출력
for page_num in pn.page_range:
    page = pn.page(page_num)
    print(page)
    for data in page:
        print(data, end = ", ")
    print()
```

# 현재 페이지(요청페이지)가 속한 page 그룹의 (page_range)에서의 시작 index와 끝 index를 조회

```python
pn.page_range[0:5]
```

```python
pn = Paginator(dataset, 5)
r = pn.page_range
r
```

```python
# 하나의 page group에 3page씩 묶는 경우
# index : 0-2, 3-5, 6-8, ...
print(r[0:3])	# 현재 page가 1, 2, 3일 때, 그 page들이 속한 page의 시작/끝 index
print(r[3:6])	# 현재 page가 4, 5, 6일 때, 그 page들이 속한 page의 시작/끝 index
print(r[7:9])	# 현재 page가 7, 8, 9일 때, 그 page들이 속한 page의 시작/끝 index
```

```python
current_page = 1 		# 현재 page 번호
page_group_count = 5	# group 당 묶을 page 수

start_idx = int((current_page-1) / page_group_count) * page_group_count
# 1,1,1,1,1, 5,5,5,5,5, 10,10,10,10,10, ...
end_idx = start_idx + page_group_count

start_idx, end_idx
```

```python
for p in pn.page_range[start_idx:end_idx]:
    print(p, end = ", ")
```
